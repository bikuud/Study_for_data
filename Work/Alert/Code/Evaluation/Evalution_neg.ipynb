```python
import pandas as pd 
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report 
```

```python
score_threshold_col=['Original','is_0.1score','is_0.2score','is_0.3score','is_0.4score','is_0.5score','is_0.6score','is_0.7score']
```

```python
# File_Path는 분류된 파일의 경로를 입력
file_path_neg=r"C:\Git_for_study\DE\Work\Neg_data\sample_neg_df_260730(2).xlsx"
scoring_soaring=pd.read_excel(file_path_neg, header=0)
soar_scoring_matrix=scoring_soaring.iloc[:,4:] # 실제 파일 구조 확인해서 몇 열까지 살릴지 고민해야함

```

```python
soar_scoring_matrix
```

```python
# Confusion_matrix 반환하는 함수
y_true = soar_scoring_matrix['AI_Label'].astype(int)
confusion_matrix_dict = {}

def matrix_extractor(df):
    for y_pred_col in score_threshold_col:
        # 해당 임계점의 예측값 가져오기
        y_pred = df[y_pred_col].astype(int)
        
        # 혼동행렬 계산
        cm = confusion_matrix(y_true, y_pred)
        tn, fp, fn, tp = cm.ravel()
        
        # 딕셔너리에 저장! (이름을 confusion_matrix_dict로 정확히 맞춤)
        confusion_matrix_dict[y_pred_col] = [tn, fp, fn, tp]
    cm_df_soar = pd.DataFrame(confusion_matrix_dict, index=['TN(정답)', 'FP(오탐)', 'FN(미탐)', 'TP(정답)'])
    return cm_df_soar
```

```python
soar_matrix=matrix_extractor(soar_scoring_matrix)
```
