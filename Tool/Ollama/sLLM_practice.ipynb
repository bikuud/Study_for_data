```python
import pandas as pd
import ollama
```

```python
df = pd.DataFrame({
    'id': [1, 2, 3],
    'review': [
        "배송도 빠르고 제품 질도 너무 마음에 듭니다! 재구매 의사 있어요.",
        "가격에 비해 성능이 너무 떨어지네요. 환불 요청합니다.",
        "그냥 일반적인 제품이에요. 특별히 나쁘지도 좋지도 않습니다."]})
```

```python
def analyze_sentiment(text):
    prompt = f"""
    다음 텍스트의 감성을 분석하여 '긍정', '부정', '중립' 중 하나로만 단답형으로 응답해줘.
    
    텍스트: "{text}"
    """
    response = ollama.chat(
        model='llama3', # 로컬에 다운로드된 모델명 입력
        messages=[
            {'role': 'system', 'content': '너는 데이터 분석을 돕는 감성 분석 전문가야.'},
            {'role': 'user', 'content': prompt}
        ],
        options={'temperature': 0} # 일관된 결과를 위해 온도를 낮춤
    )
    
    return response['message']['content'].strip()
```

```python
df['sentiment'] = df['review'].apply(analyze_sentiment)
```
